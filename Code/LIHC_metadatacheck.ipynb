```r
library(tidyverse)

```

```r
# Load the file
raw <- read.delim("LIHC.clin.merged.picked.txt", 
                   header = TRUE, 
                   row.names = 1,          # first column becomes row names
                   stringsAsFactors = FALSE)

# Transpose so patients = rows, variables = columns (normal dataframe format)
df <- as.data.frame(t(raw))

# Check it
dim(df)       # should be ~370 patients x however many variables
head(df)  # peek at first 5 columns
```

```r
# check what type R thinks each column is
str(df)

# or check a specific one
class(df$years_to_birth)     # should say "numeric"
class(df$gender)             # should say "character" or "factor"

# see unique values of a categorical variable
unique(df$pathologic_stage)  # shows you Stage I, II, III, IV
unique(df$gender)            # shows male, female
```

```r
df$years_to_birth <- as.numeric(df$years_to_birth)
df$days_to_death <- as.numeric(df$days_to_death)
df$days_to_last_followup <- as.numeric(df$days_to_last_followup)
```

```r
library(dplyr)

df$pathologic_stage <- case_when(
  df$pathologic_stage == "stage i"    ~ 1,
  df$pathologic_stage == "stage ii"   ~ 2,
  df$pathologic_stage == "stage iii"  ~ 3,
  df$pathologic_stage == "stage iiia" ~ 3,
  df$pathologic_stage == "stage iiib" ~ 3,
  df$pathologic_stage == "stage iiic" ~ 3,
  df$pathologic_stage == "stage iv"   ~ 4,
  df$pathologic_stage == "stage iva"  ~ 4,
  df$pathologic_stage == "stage ivb"  ~ 4
)
```

```r
df$gender <- as.factor(df$gender)
```

```r
str(df$years_to_birth)      # should say num
str(df$pathologic_stage)    # should say num
str(df$gender)              # should say Factor
```

```r
# continuous variables
summary(df$years_to_birth)
summary(df$days_to_death)
summary(df$days_to_last_followup)

# categorical variables
table(df$gender)
table(df$radiation_therapy)
table(df$pathologic_stage)
table(df$vital_status)
```

```r
# check these ones
table(df$tumor_tissue_site)
table(df$pathology_T_stage)
table(df$pathology_N_stage)
table(df$pathology_M_stage)
table(df$histological_type)
table(df$residual_tumor)
table(df$race)
table(df$ethnicity)

# these are numeric-ish so use summary
summary(df$days_to_last_known_alive)
summary(df$date_of_initial_pathologic_diagnosis)
```
